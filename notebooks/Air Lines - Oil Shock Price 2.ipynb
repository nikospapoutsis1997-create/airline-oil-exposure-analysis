```python
# Importing the necessary libraries.

import re
import numpy as np
import pandas as pd
from pandas.api.types import CategoricalDtype
import matplotlib.pyplot as plt
import yfinance as yf
```

```python
# Loading the data, assigning my names to tickers.

data = {}
tickers = ['RY4C.DE', 'EZJ.L', 'WIZZ.L', 'LHA.DE', 'AF.PA', 'IAG.L', 'BZ=F', 'GBPEUR=X', 'EUR=X']
my_names = ['ryanair', 'easyjet', 'wizzair', 'lufthansa', 'airfrance', 'iag', 'brent', 'gbpeur', 'usdeur']

assert len(tickers) == len(my_names)

for ticker, name in zip(tickers, my_names):
    df = yf.download(ticker, start = '2019-01-01', end = '2026-08-26')
    data[name] = df
```

```python
# Checking the structure of the dataframes.

for key, value in data.items():
    display(key, value)
```

```python
# Formatting structure of the dataframes.

def format_structure(df):
    """
    Basic formatting of the structure, resetting index, removing multiindex,
    assigning names to dataframe index.
    Dropping unnecessary columns for the analysis e.g high, low columns
    and also volume for the proxy dataframes such as brent, gbpeur and usdeur
    which are only needed for the conversion to EUR currency.
    Adding separate name columns for future use on merging.
    
    """
    for key in df.keys():
        df[key] = df[key].reset_index()
        df[key].columns = df[key].columns.droplevel(1)
        df[key].columns.name = key
        df[key]['name'] = key
        df[key].columns = df[key].columns.str.strip().str.lower()
        if key in ['brent', 'gbpeur', 'usdeur']:
            df[key] = df[key].drop(columns = ['high', 'low', 'volume'])
            df[key] = df[key][['date', 'open', 'close', 'name']]
        else:
            df[key] = df[key].drop(columns = ['high', 'low'])
            df[key] = df[key][['date', 'open', 'close', 'volume', 'name']]
        df[key] = df[key].sort_values('date')
    return df

format_structure(data)
```

```python
# Checking new structure

for key in data.keys():
    display(data[key])
```

```python
# Currency conversion

def currency_conversion(df):
    for key in df.keys():
        if key in ['easyjet', 'wizzair', 'iag']:
            df[key][['open', 'close']] = df[key][['open', 'close']] / 100 # Converting GBp to GBP
            df[key] = pd.merge(df[key], df['gbpeur'], on = 'date', how = 'left', suffixes = (f'_{key}', '_gbpeur')) # Merging for calculations
            df[key][f'open_{key}'] = df[key][f'open_{key}'] * df[key]['open_gbpeur'] # Converting open price
            df[key][f'close_{key}'] = df[key][f'close_{key}'] * df[key]['close_gbpeur'] # Converting close price
            df[key] = df[key].drop(columns = ['open_gbpeur', 'close_gbpeur', 'name_gbpeur']) # Dropping unnecessary columns
            df[key] = df[key].rename(columns = {f'open_{key}': 'open', f'close_{key}': 'close', f'name_{key}': 'name'}) # Renaming columns
            df[key].columns.name = key # Assigning column names
        elif key in ['brent']:
            df[key] = pd.merge(df[key], df['usdeur'], on = 'date', how = 'left', suffixes = (f'_{key}', '_usdeur'))
            df[key][f'open_{key}'] = df[key][f'open_{key}'] * df[key]['open_usdeur']
            df[key][f'close_{key}'] = df[key][f'close_{key}'] * df[key]['close_usdeur']
            df[key] = df[key].drop(columns = ['open_usdeur', 'close_usdeur', 'name_usdeur'])
            df[key] = df[key].rename(columns = {f'open_{key}': 'open', f'close_{key}': 'close', f'name_{key}': 'name'})
            df[key].columns.name = key
    return df

currency_conversion(data)
```

```python
# Dropping entries with 0 volume.


def drop_zero_volume_entries(df):
    for key in df.keys():
        try:
            df[key] = df[key][df[key]['volume'] > 0]
        except KeyError:
            continue
    return df

drop_zero_volume_entries(data)
```

```python
# Dropping gbpeur, usdeur frames which are no longer needed.

forex = ['gbpeur', 'usdeur']

for key in forex:
    try:
        data.pop(key, None)
    except KeyError:
        continue
```

```python
# Checking for duplicated rows and dtypes

for key in data.keys():
    display(data[key].dtypes, data[key].duplicated().value_counts(dropna = False))
```

```python
# Converting name columns to category dtype.

def to_cat(df):
    for key in df.keys():
        df[key]['name'] = df[key]['name'].astype('category')
    return df

to_cat(data)
```

```python
# Creating the final tables before loading the data to the SQL server.

def staging(df):
    airlines = []
    for key in df.keys():
        if key != 'brent':
            airlines.append(df[key])
        else:
            brent = df[key]
    combined_airlines = pd.concat(airlines, axis = 0)
    return combined_airlines, brent

airlines, brent = staging(data)
```

```python
from sqlalchemy import create_engine

server = '.'
database = 'AirlineOilProject'

connection_string = f'mssql+pyodbc://@{server}/{database}?driver=ODBC+Driver+18+for+SQL+Server&trusted_connection=yes&TrustServerCertificate=yes'

engine = create_engine(connection_string)
```

```python
airlines.to_sql('fact_airline_prices', con = engine, if_exists = 'replace', index = False)
brent.to_sql('dim_brent', con = engine, if_exists = 'replace', index = False)
```
